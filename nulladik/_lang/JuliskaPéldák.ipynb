#### A gömb térfogata
Egy $r$ sugarú gömb térfogata: $V=\frac{4r^3\pi}{3}$, ezt számolja ki a következő:

```julia
function gömbV(r)
    4*r^3*pi/3
end
println("r=1 V=",gömbV(1))
println("r=2 V=",gömbV(2))
```

A fenti térfogat-számoló függvényt megadhatjuk még az egysoros:

```julia
gömbV2(r)=4*r^3*pi/3
println("r=1 V=",gömbV2(1))
println("r=2 V=",gömbV2(2))
```

változatban is. Ezt rövidebb fv-ek definiálására használjuk. Végül van az ú.n. névtelen forma:

```julia
println("r=1 V=",((r)->4π/3*r^3)(1))
println("r=2 V=",((r)->4π/3*r^3)(2))
```

#### Másodfokú egyenlet gyökei
Az $ax^2+bx+c=0$ megoldásai $x_{1,2}=\frac{-b\pm\sqrt{b^2-4ac}}{2a}$, ami Julia-ban:

```julia
function másodfokú(a,b,c)
    d=sqrt(b^2-4a*c)
    (-b-d)/(2a),(-b+d)/(2a)
end
```

```julia
println(másodfokú(1,2,1))
println(másodfokú(1,0,-2))
```

#### Ciklus és elágazás. 
Probléma: hány szám van az $[1,100]$ intervallumban, mely $2$-vel vagy $3$-mal vagy $5$-tel osztható?

```julia
a=collect(1:100)
össz=0
for n in a
    if n%2==0 || n%3==0 || n%5==0
        println(n)
        össz=össz+1
    end
end
println("a megoldás:",össz)
```
