függvény periodikus kiterjesztésének Fourier együtthatóit "kézzel". A kapott

felhasználó közelítését.


# Megoldás

Ez a negyszögfüggvény transzformáltja, melynek ismerjük a Fourier sorát:

$$
f(x) = \frac{4}{\pi} \sum_{n=1}^{\infty} \frac{\sin((2n-1)x)}{2n-1}
$$

Ezt a függvény fogjuk transzformálni.
Kell egy lineáris függvény mely a $[-2,2]$ intervallumot leképezi a $[-\pi,\pi]$-re:

$$
x \to \frac{\pi}{2}x
$$

Ezzel a $f(\frac{\pi}{2}x)$ már a $[-2,2]$-n viselkedik úgy mint négyszög. De a mi függvényünk 
eléréséhez kellenek még külső transzformációk is:

$$
3f\left( \frac{\pi}{2}x\right)+3 = \frac{12}{\pi} \sum_{n=1}^{\infty} \frac{\sin\left(\frac{(2n-1)\pi x}{2}\right)}{2n-1} + 3
$$

# Python

A végtelen sor elejét létrehozzuk és ábrázoljuk.

```python
import numpy as np
# "N" tagig veszi és kiértékeli az "x" helyeken (partial sum)
def reszletosszeg(N, x):
   y = np.zeros(np.size(x)); # ebben lesznek az értékek
   for n in range(1,N+1):
      y+=np.sin((2*n-1)*np.pi*x*0.5)/(2*n-1)
  
   y=y*12/np.pi+3
   return y  
```

```python
import matplotlib.pyplot as plt

xx = np.linspace(-6,6,1000); # ca. 3 periódus

ax=plt.subplot(3,1,1)
yy = reszletosszeg(10, xx) # 10 tag
ax.plot(xx, yy)
plt.title("10 tag")

ax=plt.subplot(3,1,3)
yy = reszletosszeg(33, xx) # 33 tag
ax.plot(xx, yy)
plt.title("33 tag")

```
