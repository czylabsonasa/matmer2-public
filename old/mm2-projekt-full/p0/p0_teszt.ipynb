```matlab
syms y(t)
```

```matlab
sym_eq = "diff(y)+y == 2*exp(3*t)-1";


% not too general!!! only for this type
tex_eq = strrep(sym_eq, "diff(y)", "y'");
tex_eq = strrep(tex_eq, "*", "");
tex_eq = strrep(tex_eq, "exp", "e^");
tex_eq = strrep(tex_eq, "(", "{");
tex_eq = strrep(tex_eq, ")", "}");

tex_eq = strrep(tex_eq, "==", "=")
```

```matlab
eq = eval(sym_eq);


s1 = dsolve(eq, y(-3)==-1);
f1 = matlabFunction(s1);

s2 = dsolve(eq, y(-3)==0);
f2 = matlabFunction(s2);

s3 = dsolve(eq, y(-3)==1);
f3 = matlabFunction(s3);

xx = linspace(-3,3);
plot(xx, f1(xx), xx, f2(xx), xx, f3(xx))

ylim([-3, 10])
```
