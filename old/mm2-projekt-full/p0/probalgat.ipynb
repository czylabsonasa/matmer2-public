```python
from sympy import symbols, Eq, Function, dsolve
f, g = symbols("f g", cls=Function)
x = symbols("x")
eqs = [Eq(f(x).diff(x), g(x)), Eq(g(x).diff(x), f(x))]
dsolve(eqs, [f(x), g(x)])
dsolve(eqs, [f(x), g(x)], ics={f(0): 1, g(2): 3})
```
