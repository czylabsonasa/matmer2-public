### learn Mustache.jl

```julia
using Mustache
```

```julia
tpl = mt"""
Hello {{:name}}
You have just won {{:value}} dollars!
{{#:in_ca}}
Well, {{:taxed_value}} dollars, after taxes.
{{/:in_ca}}
"""
d = Dict(
:name => "Chris",
:value => 10000,
:taxed_value => 10000 - (10000 * 0.4),
:in_ca => true)

Mustache.render(tpl, d)|>println
```

```julia
function ttt(temp)
    name="Chris"
    value=10000
    taxed_value=10000 - (10000 * 0.4)
    in_ca=true
    temp(Base.@locals)
end
ttt(tpl)|>println
```
