```julia
using JSON3
using Mustache
```

```julia
p0_template=raw"""
### {{& :id}}

#### A
Vezesse le a 
$$
{{& :equation}}
$$
kde megoldását kézzel a konstans variálása módszert használva.

#### B
Oldja meg szimbolikusan a `dsolve` függvény segítségével az egyenletet. 

#### C
Ábrázolja az egyenlet vektormezejét a ${{:the_domain}}$ tartományon és rajzoltassa rá az
$$
{{& :the_ivp}}
$$
kezdeti értékekhez tartozó megoldásokat is.
""";
```

```julia
text=read("p0_feladatok.json",String)
data=copy(JSON3.read(text))
println(keys(data[:f1]))

domain_tpl=raw"[{{:xmin}},{{:xmax}}]\times[{{:ymin}},{{:ymax}}]"
ivp_tpl=raw"y({{:t1}})={{:y1}}\quad y({{:t2}})={{:y2}}\quad y({{:t3}})={{:y3}}"
out=String[]
for i in 1:16
    #data["f$(i)"]|>println
    akt=data[Symbol("f$(i)")]
    xmin,xmax=akt[:domain][1]
    ymin,ymax=akt[:domain][2]
    akt[:the_domain]=Mustache.render(domain_tpl,Base.@locals)
    t1,t2,t3=akt[:ivp][1]
    y1,y2,y3=akt[:ivp][2]
    akt[:the_ivp]=Mustache.render(ivp_tpl,Base.@locals)
    akt_str=Mustache.render(p0_template,akt) 
    push!(out,akt_str)
end
p0_feladatok_str="## p0\n\n$(join(out,'\n'))\n"
write("p0_feladatok_tmp.md",p0_feladatok_str)
```
