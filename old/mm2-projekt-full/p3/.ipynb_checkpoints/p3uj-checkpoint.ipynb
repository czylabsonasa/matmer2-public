### 3. projekt
* A javított-Euler módszer-t kell megvalósítani $10,20,50$ lépést használva, ábrázolni és az eredményt összehasonlítani az `ode23` által adott-tal.

$$
   y''+y'-ty=t+1\\

$$

$$
