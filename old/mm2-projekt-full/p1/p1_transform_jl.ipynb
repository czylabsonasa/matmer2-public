```julia
using JSON3
using Mustache
cd("../p1")
```

```julia
p1_template=raw"""
### {{& :id}}

#### A

Oldja meg az
$$
{{& :equation}}
$$
kde-t a dsolve segítségével. 

#### B
Ábrázolja az egyenlet vektormezejét a ${{:the_domain}}$ tartományon és rajzoltassa rá az
$$
{{& :the_ivp}}
$$
kezdeti értékekhez tartozó megoldásokat is.
""";
```

```julia
text=read("p1_feladatok.json",String)
data=copy(JSON3.read(text))
println(keys(data[:f1]))

domain_tpl=raw"[{{:xmin}},{{:xmax}}]\times[{{:ymin}},{{:ymax}}]"
ivp_tpl=raw"y({{:t1}})={{:y1}}\quad y({{:t2}})={{:y2}}\quad y({{:t3}})={{:y3}}"
out=String[]
for i in 1:16
    #data["f$(i)"]|>println
    akt=data[Symbol("f$(i)")]
    xmin,xmax=akt[:domain][1]
    ymin,ymax=akt[:domain][2]
    akt[:the_domain]=Mustache.render(domain_tpl,Base.@locals)
    t1,t2,t3=akt[:ivp][1]
    y1,y2,y3=akt[:ivp][2]
    akt[:the_ivp]=Mustache.render(ivp_tpl,Base.@locals)
    akt_str=Mustache.render(p1_template,akt) 
    push!(out,akt_str)
end
p1_feladatok_str="## p1\n\n$(join(out,'\n'))\n"
write("p1_feladatok_tmp.md",p1_feladatok_str)
```
