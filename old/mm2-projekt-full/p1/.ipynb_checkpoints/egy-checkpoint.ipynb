# 1. projekt - F1 - neptun kód - név


Oldja meg

$$
y'-y=\sin(x)
$$

kde-t a Matlab/Octave `dsolve` függvénye segítségével. 

Ábrázolja az egyenlet vektormezejét a $[-3,3] \times [-3,10]$-en és rajzoltassa rá az $y(-3)=2.5,\ \  y(-3)=2.7,\ \  y(-3)=3.1$
kezdeti értékekhez tartozó megoldásokat is.


## dsolve megoldás

```matlab
syms y(t)
egyenlet = diff(y) - y == sin(t)
megoldas = dsolve(egyenlet)


```

## Vektormező + megoldások

explicit alakra hozva:
$$
y' = y + \sin(t) = f(t, y)
$$

```matlab
tb = linspace(-3,3,20);
yb = linspace(-3,10,20);
[T, Y] = meshgrid(tb, yb);

DT = ones(size(T));
DY = Y + sin(T);
H = sqrt(DT.^2 + DY.^2);

DT = DT ./ H;
DY = DY ./ H;


clf;
quiver(T, Y, DT, DY)

m1 = dsolve(egyenlet, y(-3)==2.5); % danger mod! 
num1 = matlabFunction(m1);

m2 = dsolve(egyenlet, y(-3)==2.7);
num2 = matlabFunction(m2);

m3 = dsolve(egyenlet, y(-3)==3.1);
num3 = matlabFunction(m3);
hold on;

xx = linspace(-3,3);
plot(xx, num1(xx),"r")
plot(xx, num2(xx),"r")
plot(xx, num3(xx),"r")


ylim([-3.5, 10.5])
```

## a középső variálása


```matlab
yk = 0.57;

soky = linspace(yk, 10, 3333);
maxjo = 10;
for it=2:33
  yy = soky(it);
  m = dsolve(egyenlet, y(-3)==yy); 
  num = matlabFunction(m);
  vals = num(xx);
  if min(vals)<-3 || max(vals)>10
    maxjo = soky(it-1);
    break;
  end
end


soky = linspace(-3,yk, 100);
minjo = -3;
for it=100-1:-1:1
  yy = soky(it);
  m = dsolve(egyenlet, y(-3)==yy); 
  num = matlabFunction(m);
  vals = num(xx);
  if min(vals)<-3 || max(vals)>10
    minjo = soky(it+1);
    break;
  end
end
minjo
maxjo

```
