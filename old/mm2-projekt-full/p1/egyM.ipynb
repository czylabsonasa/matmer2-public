Ábrázolja az egyenlet vektormezejét a $[-3,3] \times [-3,10]$-en és rajzoltassa rá az $y(-3)=0.55,\ \  y(-3)=0.6,\ \  y(-3)=1$
kezdeti értékekhez tartozó megoldásokat is.


## dsolve megoldás

```matlab
syms y(t)
egyenlet = diff(y) - y == sin(t)
megoldas = dsolve(egyenlet)


```

## Vektormező + megoldások

explicit alakra hozva:
$$
y' = y + \sin(t) = f(t, y)
$$

```matlab
% számolások

tb = linspace(-3,3,20);
yb = linspace(-3,10,30);
[T, Y] = meshgrid(tb, yb);

DT = ones(size(T));
DY = Y + sin(T);
H = sqrt(DT.^2 + DY.^2);

DT = DT ./ H;
DY = DY ./ H;


y0 = [0.55, 0.6, 1];
m1 = dsolve(egyenlet, y(-3)==y0(1));
num1 = matlabFunction(m1);

m2 = dsolve(egyenlet, y(-3)==y0(2));
num2 = matlabFunction(m2);

m3 = dsolve(egyenlet, y(-3)==y0(3));
num3 = matlabFunction(m3);

```

```matlab
%plot -s 1000,500


quiver(T, Y, DT, DY)
hold on;

xx = linspace(-3,3);
plot(xx, num1(xx),"LineWidth",1.5)
plot(xx, num2(xx),"LineWidth",1.5)
plot(xx, num3(xx),"LineWidth",1.5)

legend(...
  "vector field", ...
  "y(-3)="+string(y0(1)), "y(-3)="+string(y0(2)), "y(-3)="+string(y0(3)), ...
  "Location", "northeastoutside");


ylim([-3.5, 10.5])
```
